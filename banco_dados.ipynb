```python
import sqlalchemy
```

```python
from sqlalchemy import create_engine, MetaData, Table, inspect
```

```python
engine = create_engine('sqlite:///:memory:')
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/refs/heads/main/clientes_banco.csv'
```

```python
import pandas as pd
```

```python
dados = pd.read_csv(url)
```

```python
dados
```

```python
dados.to_sql('clientes', engine, index=False)
```

```python
#verificando se a tabela foi criada no banco de dados
inspector = inspect(engine)
```

```python
print(inspector.get_table_names())
```

```python
#acessando o banco de dados
query = 'SELECT * FROM clientes WHERE Categoria_de_renda="Empregado"'
```

```python
empregados = pd.read_sql(query, engine)
```

```python
empregados
```

```python
empregados.to_sql('empregados', con=engine, index=False)
```

```python
#consultando tabela completa
pd.read_sql_table('empregados', engine)
```

```python
#filtrando colunas de uma tabela
pd.read_sql_table('empregados', engine, columns=['ID_Cliente', 'Grau_escolaridade', 'Rendimento_anual'])
```

```python
query = 'SELECT * FROM clientes'
```

```python
from sqlalchemy import text
#deletando algum dado
query = 'DELETE FROM clientes WHERE ID_Cliente=5008804'
with engine.connect() as conn:
  conn.execute(text(query))
```

```python
from sqlalchemy import text
# Deleting data for ID_Cliente=5008805
query = 'DELETE FROM clientes WHERE ID_Cliente=5008805'
with engine.connect() as conn:
  conn.execute(text(query))
  conn.commit() # Explicitly commit the transaction
```

After running the delete command, let's check the `clientes` table again to verify that `ID_Cliente=5008805` has been removed.

```python
# Display the clientes table after deletion
pd.read_sql_table('clientes', engine)
```

```python
#alterando dados
query = 'UPDATE clientes SET Grau_escolaridade="Ensino superior" WHERE ID_Cliente=5008808'
with engine.connect() as conn:
  conn.execute(text(query))
  conn.commit()
```
